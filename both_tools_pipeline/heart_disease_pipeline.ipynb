```python
from pathlib import Path
from IPython.display import display
import mlflow
import mlflow.sklearn
from mlflow.models import infer_signature
%load_ext noworkflow
import noworkflow.now.ipython as nip

dataset_path = Path("../datasets/heart.csv").resolve()
project_root = dataset_path.parent.parent
tracking_db_path = project_root / "mlflow.db"
tracking_uri = f"sqlite:///{tracking_db_path.as_posix()}"
mlflow.set_tracking_uri(tracking_uri)
mlflow.set_experiment("heart-disease-svm")

if mlflow.active_run() is not None:
    mlflow.end_run()

active_run = mlflow.start_run(run_name="svm_provenance")
```

# Dataset Loading 

```python
import pandas as pd
import numpy as np

dataframe = pd.read_csv("../datasets/heart.csv")
dataframe.head()
```

# Data Exploration and Analysis

```python
dataframe.describe()
```

```python
dataframe.info()
```

```python
# Setting 6th row, cholesterol as NULL, to simulate a dataset having a row with null value
dataframe.loc[5, 'Cholesterol'] = np.nan

# Checking null values in a dataset
dataframe.isnull().sum()
```

```python
# Printing 4th to 8th rows
dataframe.loc[[3,4,5,6,7]]
```

```python
# Dropping null rows (in this case, 6th row. Notice that now total rows is 917)
# inplace=True used to make the change in the dataframe itself, and not return a copy instead
dataframe.dropna(inplace = True) 
```

```python
dataframe.head(7)
```

## Plotting

```python
import matplotlib.pyplot as plt

dataframe.plot(kind='scatter', x='Cholesterol' , y='RestingBP', color='red')
```

***But wait, according to the scatterplot, there are rows with cholesterol 0. This is not realistic, and must be a problem with the dataset. Therefore, all rows with cholesterol 0 must be dropped.***

```python
for x in dataframe.index:
  if dataframe.loc[x, "Cholesterol"] == 0:
    dataframe.drop(x, inplace = True)
    print(f"Row {x} dropped")
```

```python
dataframe.info()
```

## Converting Categorical Data to Numerical Data (One Hot Encoding)

```python
dataframe.head(10)
```

```python
# One-Hot Encoding
temp_df = pd.get_dummies(dataframe["Sex"], dtype=int)
temp_df.head()
```

```python
temp_df.drop(columns=["F"],inplace=True)
dataframe.drop(columns=["Sex"], inplace=True)
dataframe = pd.concat([dataframe,temp_df], axis=1)
dataframe.rename(columns={"M":"Sex"}, inplace=True)
dataframe.head()
```

## Converting Categorical Data to Numerical Data (using Label Encoder)

**Some features have more than two categories. These categories must be replaced with 0,1,2,3... before passing to model. To achieve this, sklearn's LabelEncoder is used to replace categorical values with numerical values. The mapping is shown for each replacement.**

```python
%%now_run --name new_simulation --interactive

from sklearn.preprocessing import LabelEncoder

LE = LabelEncoder()

dataframe['ChestPainTypeLE'] = LE.fit_transform(dataframe['ChestPainType'])
dataframe['RestingECGLE'] = LE.fit_transform(dataframe['RestingECG'])
dataframe['ExerciseAnginaLE'] = LE.fit_transform(dataframe['ExerciseAngina'])
dataframe['ST_SlopeLE'] = LE.fit_transform(dataframe['ST_Slope'])
```

```python
dataframe.head()
```

```python
dataframe.drop(columns=["ChestPainType", "RestingECG", "ExerciseAngina", "ST_Slope" ] , inplace=True)
dataframe.head()
```

```python
import seaborn as sns

corr_matrix = dataframe.corr()

sns.heatmap(corr_matrix, cmap='Purples', annot=True)

plt.show()
```

# Train-Test Splitting

```python
from sklearn.model_selection import train_test_split

X = dataframe.drop(['HeartDisease'], axis=1)

y = dataframe['HeartDisease']
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.2,random_state=0)
```

# Model Training

## SVM

```python
from sklearn.svm import SVC

svm = SVC(random_state=42)

# svm = SVC(kernel="linear", C=2.0, gamma="scale", random_state=42)

# svm = SVC(kernel="", C=1.5, gamma="scale", random_state=42)

svm.fit(X_train,y_train)
y_predicted_svm = svm.predict(X_test)
print(y_predicted_svm)
```

# Evaluation

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

accuracy = accuracy_score(y_test, y_predicted_svm)
precision = precision_score(y_test, y_predicted_svm, average='macro')
recall = recall_score(y_test, y_predicted_svm, average='macro')
f1 = f1_score(y_test, y_predicted_svm, average='macro')

print("------------------------------")
print(f"Accuracy : {accuracy}")
print(f"Precision: {precision}")
print(f"Recall   : {recall}")
print(f"F1 Score : {f1}")
print("------------------------------")
```

# MLflow — Captura de Proveniência

## 2. Início da execução

```python
# Definindo tags
run_tags = {
    "pipeline": "heart_disease_pipeline",
    "provenance_scope": "data_model_metrics",
    "framework": "scikit-learn",
    "model_type": type(svm).__name__,
}
mlflow.set_tags(run_tags)
```

## 3. Proveniência dos dados

```python
# Criando a representação do dataset no MLflow
raw_dataframe = pd.read_csv(dataset_path)
mlflow_dataset = mlflow.data.from_pandas(
    raw_dataframe,
    source=str(dataset_path),
    targets="HeartDisease",
    name="heart-disease-raw",
)
mlflow.log_input(mlflow_dataset, context="raw_input")

# Salvando uma cópia completa do arquivo CSV usado pelo pipeline
mlflow.log_artifact(str(dataset_path), artifact_path="dataset")

# Reunindo as principais informações sobre a origem e o uso dos dados
data_provenance = {
    "source_path": str(dataset_path),
    "raw_rows": len(raw_dataframe),
    "raw_columns": len(raw_dataframe.columns),
    "processed_rows": len(dataframe),
    "training_rows": len(X_train),
    "test_rows": len(X_test),
    "target_column": "HeartDisease",
    "feature_columns": list(X_train.columns),
}
mlflow.log_dict(data_provenance, "provenance/data.json")

pd.Series(data_provenance, name="value").to_frame()
```

## 4. Matriz de confusão

```python
# Gerando a matriz de confusão
from sklearn.metrics import ConfusionMatrixDisplay

confusion_matrix_display = ConfusionMatrixDisplay.from_predictions(
    y_test,
    y_predicted_svm,
    display_labels=["Sem doença", "Com doença"],
    cmap="Blues",
    colorbar=False,
    values_format="d",
)
confusion_matrix_display.ax_.set_title("Matriz de confusão - SVM")
confusion_matrix_figure = confusion_matrix_display.figure_
mlflow.log_figure(confusion_matrix_figure, "plots/confusion_matrix.png")
plt.show()
```

## 5. Parâmetros, métricas e modelo

São registrados os hiperparâmetros do SVM, a divisão dos dados, as métricas já calculadas, o resumo do pré-processamento, o esquema de entrada e o modelo treinado.

```python
# Parâmetros do modelo, da divisão e dos dados
run_params = {
    "model.kernel": svm.get_params()["kernel"],
    "model.C": svm.get_params()["C"],
    "model.gamma": svm.get_params()["gamma"],
    "model.random_state": svm.get_params()["random_state"],
    "split.test_size": 0.2,
    "split.random_state": 0,
    "data.feature_count": X_train.shape[1],
}
mlflow.log_params(run_params)

# Métricas de avaliação do pipeline
run_metrics = {
    "accuracy": float(accuracy),
    "precision_macro": float(precision),
    "recall_macro": float(recall),
    "f1_macro": float(f1),
}
mlflow.log_metrics(run_metrics)

# Descrição das transformações feitas antes do treinamento
preprocessing_description = (
    "Foram removidos os registros com valores ausentes ou colesterol igual a zero. "
    "A coluna Sex foi convertida com one-hot encoding, enquanto ChestPainType, "
    "RestingECG, ExerciseAngina e ST_Slope foram codificadas com LabelEncoder."
)
mlflow.log_text(preprocessing_description, "provenance/preprocessing.txt")

model_signature = infer_signature(X_test, y_predicted_svm)
input_example = X_test.head(5)

mlflow.sklearn.log_model(
    sk_model=svm,
    name="model",
    signature=model_signature,
    input_example=input_example,
)

mlflow_run_id = active_run.info.run_id
mlflow.end_run()
print(f"MLflow run_id: {mlflow_run_id}")
```
